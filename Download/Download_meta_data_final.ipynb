```python
import pandas as pd

identifiers = [
    "pfm_multiplastic-15-150", "pfm_multiplastic-20-150", "pfm_multiplastic-30-150",
    "pfm_multiplastic-20-100", "pfm_multiplastic-30-100", "pfm_multiplastic-30-75",
    "pfm_multiplastic-20-75", "pfm_multiplastic-15-75", "pfm_multiplastic-15-100",
    "pfm_multiplastic-20-100"
]

metadata_values = []

# Loop through the identifiers and extract the metadata values for 'Power' and 'Scan Speed'
for identifier in identifiers:
    record = manager.record(identifier=identifier)
    process_params = record.get_metadatum("Process Parameters")  # Assuming metadata is stored under 'Process Parameters'
    
    power = None
    power_unit = None
    scan_speed = None
    scan_speed_unit = None
    
    if process_params:
        for param in process_params['value']:
            if param['key'] == 'Power':
                power = param['value']
                power_unit = param['unit']
            elif param['key'] == 'Scan Speed':
                scan_speed = param['value']
                scan_speed_unit = param['unit']
    
    metadata_values.append({
        'Power': power,
        'Power Unit': power_unit,
        'Scan Speed': scan_speed,
        'Scan Speed Unit': scan_speed_unit
    })

# Create a DataFrame from the extracted metadata values
df = pd.DataFrame(metadata_values)
print(df)
```
